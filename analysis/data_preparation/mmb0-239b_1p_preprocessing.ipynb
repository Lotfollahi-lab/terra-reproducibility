```python
import requests
import os
```

```python
url = "https://datasets.cellxgene.cziscience.com/90cdd3f7-e61e-43ed-97e9-ddc0f7e87827.h5ad"
out_path = "/nfs/team361/sb75/DATASETS/bronze/mmb0-239b_1p/90cdd3f7-e61e-43ed-97e9-ddc0f7e87827.h5ad"

# make sure directory exists
os.makedirs(os.path.dirname(out_path), exist_ok=True)

with requests.get(url, stream=True) as r:
    r.raise_for_status()
    with open(out_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=8192):
            if chunk:
                f.write(chunk)

print("Download complete!")
```
