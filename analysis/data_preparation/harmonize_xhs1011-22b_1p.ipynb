```python
import os
import scanpy as sc

#load_folder_path = '/lustre/scratch126/cellgen/lotfollahi/DATASETS/silver/xhs1011-22b_1p'
load_folder_path = '/nfs/team361/sb75/DATASETS/silver/xhs1011-22b_1p'

for subdir, _, files in os.walk(load_folder_path):
    print(f'Loading AnnData objects from {subdir}.')
    for file in files:
        if file.endswith('.h5ad'):
            file_path = os.path.join(subdir, file)
            adata = sc.read_h5ad(file_path)
            sample_id = adata.obs['sample_id'].unique().tolist()[0]
            if sample_id not in list(eczema_samples.keys()):
                adata.uns['condition'] = 'psoriasis'
            else:
                adata.uns['condition'] = 'eczema'
            adata.write(file_path)
            print(sample_id)
            print(adata.uns['batch'])
            print(adata.uns['condition'])
            print(adata.obs['Responder'][0])
            print(adata.obs['Site_status'][0])
            print('===')
            
```
