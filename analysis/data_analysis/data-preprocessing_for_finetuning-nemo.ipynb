## Filtering Data

```python
def filter_dataset_by_cell_ids(
    dataset,
    target_cell_ids,
    batch_size=10000,
    num_proc=4,
    temp_dir=None,
):
    """
    Filter a Hugging Face dataset to keep only cells matching a given list of cell IDs.
    
    Uses a directory to avoid permission issues when filtering large datasets.

    Parameters:
    -----------
    dataset : Dataset
        Hugging Face dataset with 'cell_id' column
    target_cell_ids : list
        List of complete cell IDs to filter for (e.g., ['1000_batch1_0', '1000_batch1_1', '1002_batch3_5'])
    batch_size : int, default=10000
        Batch size for processing
    num_proc : int, default=4
        Number of processes for parallel processing
    temp_dir : str, optional
        Custom temporary directory path. If None, uses current working directory.
        
    Returns:
    --------
    Dataset
        Filtered dataset containing only matching items.
    """    
    # Convert to set for faster lookup
    target_cell_ids_set = set(target_cell_ids)
    
    # Define the filter function
    def filter_function(ids):
        return [cell_id in target_cell_ids_set for cell_id in ids]
    
    # Cleanup function
    def cleanup_cache_file(cache_file_path):
        """Clean up cache file."""
        try:
            if os.path.exists(cache_file_path):
                os.remove(cache_file_path)
            # Also clean up the cache directory if it's empty
            cache_dir = os.path.dirname(cache_file_path)
            if os.path.exists(cache_dir) and not os.listdir(cache_dir):
                os.rmdir(cache_dir)
        except Exception:
            pass  # Silently fail if cleanup doesn't work

    # Determine temp directory
    if temp_dir is None:
        temp_dir = os.getcwd()
    
    # Create directory if it doesn't exist
    os.makedirs(temp_dir, exist_ok=True)
    cache_file = os.path.join(temp_dir, "filtered_dataset_cache_by_cell_ids.arrow")
    
    try:
        # Apply the filter with cache location
        filtered_dataset = dataset.filter(
            filter_function,
            input_columns="cell_id",
            batched=True,
            batch_size=batch_size,
            num_proc=num_proc,
            cache_file_name=cache_file,
        )
        
        # Clean up cache file after successful filtering
        cleanup_cache_file(cache_file)
        
        return filtered_dataset
    except Exception as e:
        # Clean up on error
        cleanup_cache_file(cache_file)
        raise
```

```python
def filter_dataset_by_dataset_batch(
    dataset,
    target_dataset_batch_ids,
    batch_size=10000,
    num_proc=4,
    temp_dir=None,
):
    """
    Filter a Hugging Face dataset to keep only cells matching a given dataset ID and batch IDs.
    
    Uses a directory to avoid permission issues when filtering large datasets.

    Parameters:
    -----------
    dataset : Dataset
        Hugging Face dataset with 'cell_id' column
    target_dataset_batch_ids : list
        List of dataset id and batch id to filter for (e.g., ['1000_batch1', '1000_batch2'])
    batch_size : int, default=10000
        Batch size for processing
    num_proc : int, default=4
        Number of processes for parallel processing
    temp_dir : str, optional
        Custom temporary directory path. If None, uses current working directory.
        
    Returns:
    --------
    Dataset
        Filtered dataset containing only matching items.
    """    
    # Define the filter function
    def filter_function(ids):
        return [i.split('_')[0] + '_' + i.split('_')[1] in target_dataset_batch_ids for i in ids]
    
    # Cleanup function
    def cleanup_cache_file(cache_file_path):
        """Clean up cache file."""
        try:
            if os.path.exists(cache_file_path):
                os.remove(cache_file_path)
            # Also clean up the cache directory if it's empty
            cache_dir = os.path.dirname(cache_file_path)
            if os.path.exists(cache_dir) and not os.listdir(cache_dir):
                os.rmdir(cache_dir)
        except Exception:
            pass  # Silently fail if cleanup doesn't work

    # Determine temp directory
    if temp_dir is None:
        temp_dir = os.getcwd()
    
    # Create directory if it doesn't exist
    os.makedirs(temp_dir, exist_ok=True)
    cache_file = os.path.join(temp_dir, "filtered_dataset_cache.arrow")
    
    try:
        # Apply the filter with cache location
        filtered_dataset = dataset.filter(
            filter_function,
            input_columns="cell_id",
            batched=True,
            batch_size=batch_size,
            num_proc=num_proc,
            cache_file_name=cache_file,
        )
        
        # Clean up cache file after successful filtering
        cleanup_cache_file(cache_file)
        
        return filtered_dataset
    except Exception as e:
        # Clean up on error
        cleanup_cache_file(cache_file)
        raise
```

```python
def collect_adata_from_dataset_batches(load_folder_path: str,
                                       dataset_batch_ids: list[str],
                                       cell_ids: list[str] | None = None,
                                       obs_cols: list[str] | None = None,
                                       uns_cols: list[str] | None = None,
                                       include_gene_panel_size: bool = True,
                                       exclude_folders: list[str] | None = None,
                                       ) -> ad.AnnData:
    """
    Loop through folder, read `.h5ad` files from subdirectories matching
    the provided "<dataset-id>_<batch-id>" strings, and concatenate them into
    one AnnData object.

    Files are matched by checking `adata.uns['dataset_id']` and `adata.uns['batch']`
    against the provided dataset_batch_ids. Subdirectories are pre-filtered by
    dataset-id presence in folder name (e.g., "1022" matches "xhs1022-1-55b_1p").
    
    Note: A dataset ID can match multiple subdirectories (e.g., dataset 1022 split
    across xhs1022-1-55b_1p, xhs1022-2-55b_1p, xhs1022-3-55b_1p), and data from
    all matching folders will be collected.

    Parameters
    ----------
    load_folder_path:
        Directory which is searched for AnnData objects.
    dataset_batch_ids:
        List of strings of the form "<dataset-id>_<batch-id>" (e.g., "1022_batch1").
        A subdirectory is included if it contains the dataset-id. Files are loaded
        and then filtered by checking `adata.uns['dataset_id']` and `adata.uns['batch']`
        match the dataset_id and batch_id.
    cell_ids:
        IDs of cells which are included in the final object. If `None`,
        all cells are included.
    obs_cols:
        Columns in `adata.obs` that should be retained in the final object.
        If `None`, all columns are retained.
    uns_cols:
        Keys in `adata.uns` that should be retained in the final object.
        If `None`, all keys are retained.
    include_gene_panel_size:
        If True, add 'gene_panel_size' to `adata.obs`.
    exclude_folders:
        List of folder names to skip entirely. If a folder name (or any parent
        folder in its path) matches any string in this list, it will be ignored.
        Useful for excluding specific datasets like ["hst_corpus_110m"].

    Returns
    -------
    ad.AnnData:
        The concatenated AnnData object.
    """
    if not dataset_batch_ids:
        raise ValueError("`dataset_batch_ids` must be a non-empty list of '<dataset-id>_<batch-id>' strings.")

    # Build sets of (dataset_id, batch_id) tuples for fast lookup
    desired_dataset_batches = set()
    dataset_ids_for_folder_matching = set()
    
    for db_id in dataset_batch_ids:
        if '_' not in db_id:
            raise ValueError(f"Invalid dataset_batch_id format: '{db_id}'. Expected format: '<dataset-id>_<batch-id>'")
        dataset_id, batch_id = db_id.split('_', 1)
        desired_dataset_batches.add((dataset_id, batch_id))
        dataset_ids_for_folder_matching.add(dataset_id)

    adata_list = []
    
    # Convert exclude_folders to set for faster lookup
    exclude_set = set(exclude_folders) if exclude_folders else set()

    # Walk through the load folder path and read files
    for subdir, _, files in os.walk(load_folder_path):
        # Check if any part of the path should be excluded
        if exclude_set:
            path_parts = subdir.split(os.sep)
            if any(excluded_folder in path_parts for excluded_folder in exclude_set):
                continue
        
        subdir_name = os.path.basename(subdir)
        
        # Find if any dataset_id matches this subdir
        # A dataset_id can match multiple subdirs (e.g., "1022" matches xhs1022-1-55b_1p, xhs1022-2-55b_1p, etc.)
        matching_dataset_id = None
        for dataset_id in dataset_ids_for_folder_matching:
            if dataset_id in subdir_name:
                matching_dataset_id = dataset_id
                break
        
        if matching_dataset_id is None:
            continue
        
        print(f'Scanning {subdir} for dataset_id={matching_dataset_id}...')
        
        # Load all .h5ad files and filter by dataset_id and batch_id from uns
        for file in files:
            if not file.endswith('.h5ad'):
                continue
            
            file_path = os.path.join(subdir, file)
            try:
                adata = sc.read_h5ad(file_path)
            except Exception as e:
                print(f'Warning: Could not read {file_path}: {e}')
                continue
            
            # Check if dataset_id and batch_id in uns match any desired combination
            if 'dataset_id' not in adata.uns or 'batch' not in adata.uns:
                print(f'Warning: {file_path} missing dataset_id or batch in uns, skipping.')
                del adata
                continue
            
            file_dataset_id = str(adata.uns['dataset_id'])
            file_batch_id = str(adata.uns['batch'])
            
            if (file_dataset_id, file_batch_id) not in desired_dataset_batches:
                del adata
                continue
            
            print(f'  Loading {file} (dataset_id={file_dataset_id}, batch={file_batch_id})')
            
            if cell_ids:
                adata = adata[adata.obs['cell_id'].isin(cell_ids)]
            if len(adata) == 0:
                del adata
                continue
            
            # Handle obs_cols: if None, keep all; otherwise filter to specified columns
            if obs_cols is not None:
                # Ensure 'cell_id' is included if it exists
                cols_to_keep = ['cell_id'] if 'cell_id' in adata.obs.columns else []
                cols_to_keep += [col for col in obs_cols if col in adata.obs.columns]
                adata.obs = adata.obs[cols_to_keep]
            # If obs_cols is None, keep all obs columns (no filtering)
            
            # Handle uns_cols: if None, copy all uns keys to obs; otherwise only specified ones
            if uns_cols is None:
                # Copy all uns keys to obs
                for key in adata.uns.keys():
                    adata.obs[key] = adata.uns[key]
            else:
                # Only copy specified uns keys
                for col in uns_cols:
                    if col in adata.uns:
                        adata.obs[col] = adata.uns[col]
            
            if include_gene_panel_size:
                adata.obs['gene_panel_size'] = len(adata.var_names)
            adata_list.append(adata)

    if len(adata_list) == 0:
        raise ValueError("No matching AnnData files found for the given dataset_batch_ids.")

    concatenated_adata = ad.concat(adata_list, join='outer', index_unique=None)
    
    return concatenated_adata
```

## Plotting

```python
def plot_distribution_cells_vs_patients_niche(adata):
    # Aggregate both metrics
    niche_summary = (
        adata.obs.groupby('niche19')
        .agg(
            num_patients=('Patient ID', 'nunique'),
            num_cells=('Patient ID', 'size')
        )
        .sort_values('num_cells', ascending=False)
        .reset_index()
    )

    # ---- plotting ----
    fig, ax1 = plt.subplots(figsize=(10, 8))

    order = niche_summary['niche19']

    # Left x-axis: number of patients
    sns.barplot(
        data=niche_summary,
        y='niche19',
        x='num_patients',
        order=order,
        color='steelblue',
        ax=ax1,
        label='Patients'
    )
    ax1.set_xlabel('Number of Patients')
    ax1.set_ylabel('niche19 Label')
    ax1.set_title(f"Number of Patients and Cells per niche19 Label ({niche_summary['num_cells'].sum():,} Cells)")

    # Make x-axis integer ticks from 0 → max patients
    max_patients = niche_summary['num_patients'].max()
    ax1.set_xticks(np.arange(0, max_patients + 1, 1))
    ax1.set_xlim(0, max_patients + 0.5)

    # Right x-axis: number of cells
    ax2 = ax1.twiny()
    sns.barplot(
        data=niche_summary,
        y='niche19',
        x='num_cells',
        order=order,
        color='lightcoral',
        ax=ax2,
        alpha=0.6,
        label='Cells'
    )
    ax2.set_xlabel('Number of Cells')

    plt.tight_layout()
    plt.show()
```

```python
def extract_dataset_and_batch_from_anndata(adata, id_col='cell_id'):
    """
    Extracts 'dataset_id' and 'batch' from adata.obs[id_col] strings
    like '1010_batch1_0', and adds them as new columns to adata.obs.

    Parameters
    ----------
    adata : anndata.AnnData
        The AnnData object containing obs with a cell_id column.
    id_col : str, default='cell_id'
        Column name in adata.obs containing identifiers like '1010_batch1_0'.

    Returns
    -------
    adata : anndata.AnnData
        The same AnnData object with new obs columns:
        - 'dataset_id' (int)
        - 'batch' (str)
    """
    if id_col not in adata.obs.columns:
        raise KeyError(f"Column '{id_col}' not found in adata.obs")

    # Split the ID string
    split_cols = adata.obs[id_col].astype(str).str.split('_', expand=True)

    # Assign new columns
    adata.obs['dataset_id'] = split_cols[0].astype(int)
    adata.obs['batch'] = split_cols[1]

    return adata

xhs_merge = extract_dataset_and_batch_from_anndata(xhs_merge)
xhs_merge
```

### Fix Patient IDs in Merged Adata

```python
PATIENT_RE = re.compile(r'(?:BK|CE)\d{1,3}', flags=re.IGNORECASE)

def extract_sanger_ids_concat(cell_id) -> str:
    """
    Extract all substrings like BK1, BK25, CE3, CE45, CE123 from a string.
    Concatenate them with '-' if multiple are distinct.
    If all matches are the same, keep only one.
    Returns 'NA' if no match is found.
    """
    if not isinstance(cell_id, str):
        cell_id = str(cell_id)
    matches = [m.upper() for m in PATIENT_RE.findall(cell_id)]
    
    if not matches:
        return 'NA'
    
    # if multiple matches but all identical, keep one
    unique_matches = list(dict.fromkeys(matches))  # preserves order, removes dups
    if len(set(unique_matches)) == 1:
        return unique_matches[0]
    else:
        return '-'.join(unique_matches)
```

```python
# Apply to obs index
xhs_merge.obs['Sanger patient ID from Index'] = [
    extract_sanger_ids_concat(idx) for idx in xhs_merge.obs.index.copy()
]

xhs_merge
```

```python
def assign_patient_id(row):
    """
    Create a new 'Patient ID' column using the following rules:
    1. Use 'Sanger patient ID' if it is not NaN.
    2. Otherwise, use 'Sanger patient ID from Index' 
       *unless* it equals 'BK21-BK51' or 'BK21-CE3'.
    3. If neither condition applies, return NaN.
    """
    sanger_id = row.get('Sanger patient ID')
    index_id = row.get('Sanger patient ID from Index')

    # If Sanger patient ID exists, use it
    if pd.notna(sanger_id):
        return sanger_id
    
    # Otherwise, use the index-derived one if it's valid
    if pd.notna(index_id) and index_id not in ["BK21-BK51", "BK21-CE3"]:
        return index_id
    
    # Fallback
    return pd.NA
```

```python
xhs_merge.obs['Patient ID'] = xhs_merge.obs.apply(assign_patient_id, axis=1)
xhs_merge
```
