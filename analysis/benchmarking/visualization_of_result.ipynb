```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
models = ['cellcharter', 'banksy','random-emb-average(768)','random-emb-weight-average(768)','Niche-jepa weight','Niche-jepa','Niche-justpos']
tasks = ['f1_test']

accuracy = [
    [0.8821415676],  # Model 1
    [0.9091410444],  # Model 2
    [0.7215112528456838],   # Model 4
    [0.9458],   # Model 5
    [0.9402],    # Model 6
    [0.8446203101135524],    # Model 6
    [0.1533],    # Model 6
]
'''
accuracy = [
    [0.8543716025],  # Model 1
    [0.5549141491],  # Model 2
    [0.356],   # Model 4
    [0.476],   # Model 5
    [0.5665],    # Model 6
    [0.488]    # Model 6
]
'''
# Number of models and tasks
n_models = len(models)
n_tasks = len(tasks)

# Positions of the bars on the x-axis
x = np.arange(n_tasks)

# Width of a bar
width = 0.12

# Create a bar plot with a larger figure size
fig, ax = plt.subplots(figsize=(12, 6))
for i in range(n_models):
    ax.bar(x + i * width, accuracy[i], width, label=models[i])

# Add labels, title, and legend
ax.set_xlabel('Tasks')
ax.set_ylabel('Accuracy')
ax.set_title('Accuracy of Models in Different Tasks with 4neb knn')
ax.set_xticks(x + width / 2 * (n_models - 1))
ax.set_xticklabels(tasks)

# Place legend outside the plot
ax.legend(bbox_to_anchor=(1, 1), loc='upper left')

# Adjust layout to make room for the legend
plt.tight_layout(rect=[0, 0, 0.85, 1])

# Show the plot
plt.show()
```
