```python
import os
import anndata as ad
```

```python
def process_embryo_sample(adata, sample_name, output_dir):
    print(f"Processing {sample_name}...")

    adata.var['ensembl_id'] = adata.var.index
    if 'SYMBOL' in adata.var.columns:
        adata.var.index = adata.var['SYMBOL']
        adata.var.index.name = 'gene_name'
    adata.var = adata.var[['ensembl_id']]

    # Extract spatial coordinates in microns
    adata.obsm['spatial'] = adata.obsm['xyz_um'][:, :2].copy()

    # Filter .uns to keep only keys containing the sample_name
    adata.uns = {key: value for key, value in adata.uns.items() if sample_name in key}

    print(f"  Shape: {adata.shape}, Var columns: {list(adata.var.columns)}, Obs columns: {len(adata.obs.columns)}")

    # Save as h5ad format
    output_file = os.path.join(output_dir, f'{sample_name}_norm.h5ad')
    adata.write_h5ad(output_file, compression='gzip')
    print(f"  Saved to: {output_file}")
```

```python
input_file = '/lustre/scratch126/cellgen/lotfollahi/zh4/NicheJEPA/Data/Visium_Embryo_Figure/Raw Data/adata_CS17.h5ad'
output_dir = '/lustre/scratch126/cellgen/lotfollahi/zh4/NicheJEPA/Data/Visium_Embryo_Figure/Raw Data'

# Load the big adata
adata = ad.read_h5ad(input_file)
```

```python
# Get unique sample names
sample_names = adata.obs['sample'].unique()
print(f'Found {len(sample_names)} unique samples')
print(f'Sample names: {list(sample_names)[:10]}...')   
```

```python
# Split and process each sample
for sample_name in sample_names:
    # Filter data for this sample
    sample_adata = adata[adata.obs['sample'] == sample_name].copy()
    
    # Process and save
    process_embryo_sample(sample_adata, sample_name, output_dir)

print(f'\nCompleted! Processed and saved {len(sample_names)} samples to {output_dir}')
```
