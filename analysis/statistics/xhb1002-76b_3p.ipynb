```python
import os
from collections import Counter

def get_unique_file_counts(directory):
    # Get a list of files in the directory
    files = os.listdir(directory)

    # Extract the first substring before the first '-' or '_'
    donor_substrings = [file.split('-')[0].split('_')[0] for file in files if file.startswith('AT')]
    
    # Count the unique substrings
    filecount_by_donor = Counter(donor_substrings)

    # Sort the counts by the number in the donor ID
    donors = sorted(filecount_by_donor.keys(), key=lambda x: int(''.join(filter(str.isdigit, x))))
    
    filecount_by_donor = dict(zip(donors, [filecount_by_donor[donor] for donor in donors]))
    files_by_donor = {donor: [file for file in files if file.startswith(donor)] for donor in donors}

    return filecount_by_donor, files_by_donor

# Example usage
directory = '/lustre/scratch126/cellgen/team361/DATASETS/silver/xhb1002-76b_3p'
filecount_by_donor, files_by_donor = get_unique_file_counts(directory)
for donor in files_by_donor:
    print(donor)
    for filename in files_by_donor[donor]:
        print(f'{filename}')
    print()
```

```python
directory = '/lustre/scratch126/cellgen/team361/DATASETS/silver/xhb1002-76b_3p'
silver_data_dir = '/lustre/scratch126/cellgen/team361/DATASETS/silver/'

for donor in files_by_donor:
    dataset_name = f"xhb1002-{donor}-{filecount_by_donor[donor]}b_1p"
    print(dataset_name)
    dataset_dir = os.path.join(silver_data_dir, dataset_name)
    os.makedirs(dataset_dir, exist_ok=True)
    
    for filename in files_by_donor[donor]:
        original_filepath = os.path.join(directory, filename)
        print(f"Original: {original_filepath}")
        
        new_filepath = os.path.join(dataset_dir, filename)
        print(f"New: {new_filepath}")
        
        os.system(f'cp {original_filepath} {new_filepath}')
```
